# Simulasi Potensial Listrik Dipol

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

def potential_dipole(x, y, q, d=0.1):
    """Potensial dari dipol dengan muatan +q di (d,0) dan -q di (-d,0)"""
    k = 8.99e9
    r1 = np.sqrt((x - d)**2 + y**2)
    r2 = np.sqrt((x + d)**2 + y**2)
    V = k * q / r1 - k * q / r2
    # Hindari singularitas
    V = np.nan_to_num(V, nan=0.0, posinf=0.0, neginf=0.0)
    return V

# Grid
x = np.linspace(-0.5, 0.5, 100)
y = np.linspace(-0.5, 0.5, 100)
X, Y = np.meshgrid(x, y)

# Parameter
q = 1e-9  # C
d = 0.05   # m

# Potensial
V = potential_dipole(X, Y, q, d)

# Medan listrik dari gradien potensial
Ex, Ey = np.gradient(-V, x[1]-x[0], y[1]-y[0])

# Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Kontur potensial
ax1.contour(X, Y, V, levels=20, cmap='viridis')
ax1.set_xlabel('x (m)')
ax1.set_ylabel('y (m)')
ax1.set_title('Potensial Listrik Dipol')
ax1.set_aspect('equal')
ax1.grid(True, linestyle='--', alpha=0.3)

# Vektor medan
ax2.streamplot(X, Y, Ex, Ey, color='blue', linewidth=1, density=1.5)
ax2.scatter([d, -d], [0, 0], s=100, color=['red', 'blue'], zorder=5)
ax2.set_xlabel('x (m)')
ax2.set_ylabel('y (m)')
ax2.set_title('Medan Listrik Dipol')
ax2.set_aspect('equal')
ax2.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()

# Potensial sepanjang sumbu x
x_axis = np.linspace(-0.3, 0.3, 100)
V_axis = potential_dipole(x_axis, 0, q, d)

plt.figure(figsize=(8, 4))
plt.plot(x_axis, V_axis, 'b-', linewidth=2)
plt.xlabel('x (m)')
plt.ylabel('V (V)')
plt.title('Potensial Sepanjang Sumbu x')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(y=0, color='black', linewidth=0.5)
plt.show()
```
