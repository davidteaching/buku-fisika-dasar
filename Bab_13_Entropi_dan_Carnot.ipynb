# Simulasi Entropi dan Mesin Carnot

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```python
import numpy as np
import matplotlib.pyplot as plt

# Fungsi efisiensi Carnot
def carnot_efficiency(T_h, T_c):
    return 1 - T_c / T_h

# Rentang suhu
T_h = np.linspace(300, 1000, 100)  # reservoir panas (K)
T_c = 300  # reservoir dingin (K)

eta = carnot_efficiency(T_h, T_c)

# Plot
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(T_h, eta * 100, 'b-', linewidth=2)
plt.xlabel('Suhu Reservoir Panas (K)')
plt.ylabel('Efisiensi Carnot (%)')
plt.title('Efisiensi Carnot vs T_h')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axvline(x=500, color='r', linestyle='--', alpha=0.5)
plt.axhline(y=40, color='r', linestyle='--', alpha=0.5)

# Perubahan entropi untuk perpindahan kalor spontan
def entropy_transfer(Q, T_h, T_c):
    dS_h = -Q / T_h
    dS_c = Q / T_c
    return dS_h + dS_c

Q = 1000  # J
T_h2 = 500
T_c2 = 300
dS_total = entropy_transfer(Q, T_h2, T_c2)
print(f"Perubahan entropi total untuk Q={Q} J dari {T_h2} K ke {T_c2} K: {dS_total:.2f} J/K")

# Siklus Carnot (diagram PV sederhana)
plt.subplot(1, 3, 2)
V = np.linspace(1, 4, 100)
# Isotermal T_h
P_h = 1.0 / V
# Adiabatik (sketsa)
P_adi = 0.5 / V**1.4

plt.plot(V, P_h, 'r-', linewidth=2, label='Isotermal T_h')
plt.plot(V, P_adi, 'b-', linewidth=2, label='Adiabatik')
plt.xlabel('Volume (arb. unit)')
plt.ylabel('Tekanan (arb. unit)')
plt.title('Siklus Carnot (Sketsa)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

# Entropi sebagai fungsi suhu untuk gas ideal
plt.subplot(1, 3, 3)
T = np.linspace(100, 500, 100)
S = np.log(T)  # sketsa: S ~ C_v ln T
plt.plot(T, S, 'g-', linewidth=2)
plt.xlabel('Suhu (K)')
plt.ylabel('Entropi (arb. unit)')
plt.title('Entropi Gas Ideal vs Suhu')
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
