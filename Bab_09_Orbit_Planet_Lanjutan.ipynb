# Simulasi Orbit Planet dengan Metode Numerik (Lanjutan)

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
# Kode ini sama dengan Bab 05, tetapi disertakan untuk kelengkapan.
# Gunakan notebook dari Bab 05 untuk simulasi orbit.
import numpy as np
import matplotlib.pyplot as plt

G = 6.67e-11
M_sun = 2.0e30
dt = 3600 * 6
t_total = 365 * 24 * 3600

x0, y0 = 1.5e11, 0.0
vx0, vy0 = 0.0, 29783.0

t = np.arange(0, t_total, dt)
x = np.zeros_like(t)
y = np.zeros_like(t)
vx = np.zeros_like(t)
vy = np.zeros_like(t)

x[0], y[0] = x0, y0
vx[0], vy[0] = vx0, vy0

for i in range(len(t)-1):
    r = np.sqrt(x[i]**2 + y[i]**2)
    ax = -G * M_sun * x[i] / r**3
    ay = -G * M_sun * y[i] / r**3
    vx[i+1] = vx[i] + ax * dt
    vy[i+1] = vy[i] + ay * dt
    x[i+1] = x[i] + vx[i+1] * dt
    y[i+1] = y[i] + vy[i+1] * dt

plt.figure(figsize=(8,8))
plt.plot(x, y, 'b-', linewidth=1.5)
plt.plot(0, 0, 'yo', markersize=10)
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Orbit Bumi')
plt.axis('equal')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
