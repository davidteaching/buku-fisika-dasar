# Simulasi Relativitas Khusus

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

def lorentz_factor(v):
    """Menghitung faktor Lorentz"""
    return 1 / np.sqrt(1 - (v/3e8)**2)

def time_dilation(t0, v):
    """Dilatasi waktu"""
    return t0 * lorentz_factor(v)

def length_contraction(L0, v):
    """Kontraksi panjang"""
    return L0 / lorentz_factor(v)

def relativistic_energy(m, v):
    """Energi total relativistik"""
    return lorentz_factor(v) * m * (3e8)**2

# Rentang kecepatan
v = np.linspace(0, 2.9e8, 100)
gamma = lorentz_factor(v)

# Plot
plt.figure(figsize=(12, 10))

# Faktor Lorentz
plt.subplot(2, 2, 1)
plt.plot(v/3e8, gamma, 'b-', linewidth=2)
plt.xlabel('v/c')
plt.ylabel(r'$\gamma$')
plt.title('Faktor Lorentz vs Kecepatan')
plt.grid(True, linestyle='--', alpha=0.3)
plt.axhline(y=1, color='gray', linestyle=':', alpha=0.5)

# Dilatasi waktu
plt.subplot(2, 2, 2)
t0 = 1.0  # 1 detik
plt.plot(v/3e8, time_dilation(t0, v), 'r-', linewidth=2)
plt.xlabel('v/c')
plt.ylabel('Waktu teramati (s)')
plt.title('Dilatasi Waktu ($\Delta t_0 = 1$ s)')
plt.grid(True, linestyle='--', alpha=0.3)

# Kontraksi panjang
plt.subplot(2, 2, 3)
L0 = 1.0  # 1 meter
plt.plot(v/3e8, length_contraction(L0, v), 'g-', linewidth=2)
plt.xlabel('v/c')
plt.ylabel('Panjang teramati (m)')
plt.title('Kontraksi Panjang ($L_0 = 1$ m)')
plt.grid(True, linestyle='--', alpha=0.3)

# Energi relativistik
plt.subplot(2, 2, 4)
m = 1.0  # 1 kg
E = relativistic_energy(m, v)
plt.plot(v/3e8, E/1e16, 'm-', linewidth=2)
plt.xlabel('v/c')
plt.ylabel('Energi ($10^{16}$ J)')
plt.title('Energi Relativistik untuk $m = 1$ kg')
plt.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()

# Tampilkan beberapa nilai karakteristik
v_vals = [0, 0.5, 0.9, 0.99, 0.999]
print("v/c	gamma	t_dilatasi (t0=1s)	L_kontraksi (L0=1m)")
print("-" * 70)
for v in v_vals:
    v_ms = v * 3e8
    gamma_val = lorentz_factor(v_ms)
    t_dil = time_dilation(1, v_ms)
    L_kont = length_contraction(1, v_ms)
    print(f"{v:.3f}	{gamma_val:.2f}	{t_dil:.4f} s			{L_kont:.4f} m")
```
