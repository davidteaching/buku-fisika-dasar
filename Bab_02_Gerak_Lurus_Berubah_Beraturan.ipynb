# Simulasi Gerak Lurus Berubah Beraturan

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 5, 100)
v0 = 10      # kecepatan awal (m/s)
a = 2        # percepatan (m/s^2)

x = v0 * t + 0.5 * a * t**2

plt.plot(t, x, 'b-', linewidth=2)
plt.xlabel('Waktu (s)')
plt.ylabel('Posisi (m)')
plt.title('Grafik x(t) untuk GLBB')
plt.grid(True)
plt.show()
```
