# Simulasi Proses Termodinamika Gas Ideal

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

# Konstanta
n = 1.0          # jumlah mol
R = 8.314        # J/(mol K)

# Proses Isotermal (T = 300 K)
T = 300
V = np.linspace(0.5, 5.0, 100)
P_iso = n * R * T / V

# Proses Adiabatik (gamma = 1.4 untuk gas diatomik)
gamma = 1.4
P0 = n * R * T / 2.0  # tekanan pada V=2 m^3
V_adi = np.linspace(0.5, 5.0, 100)
P_adi = P0 * (2.0 / V_adi)**gamma

# Proses Isobarik (P = 2 atm ~ 2e5 Pa)
P_isobar = 2e5
V_isobar = np.linspace(0.5, 5.0, 100)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(V, P_iso, 'b-', linewidth=2, label='Isotermal (T = 300 K)')
plt.plot(V_adi, P_adi, 'r-', linewidth=2, label='Adiabatik')
plt.axhline(y=P_isobar, color='g', linestyle='-', linewidth=2, label='Isobarik')
plt.axvline(x=2.0, color='orange', linestyle='-', linewidth=2, label='Isokhorik')

plt.xlabel('Volume (m$^3$)', fontsize=12)
plt.ylabel('Tekanan (Pa)', fontsize=12)
plt.title('Proses Termodinamika Gas Ideal', fontsize=14)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.yscale('log')  # skala log untuk melihat perbedaan lebih jelas
plt.xlim(0.5, 5.0)
plt.show()

# Hitung kerja untuk proses isotermal dari V=1 ke V=4
V1, V2 = 1.0, 4.0
W_iso = n * R * T * np.log(V2/V1)
print(f"Kerja isotermal (V1={V1}, V2={V2}): {W_iso:.2f} J")
```
