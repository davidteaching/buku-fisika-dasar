# Simulasi Lensa Cembung dan Bayangan

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

def lens_image(f, s, h=1.0):
    """Menghitung posisi dan sifat bayangan lensa tipis"""
    if s == f:
        return None, None, None  # Bayangan di tak hingga
    s_prime = 1 / (1/f - 1/s)
    M = -s_prime / s
    h_prime = M * h
    return s_prime, M, h_prime

# Parameter
f = 10  # cm

# Posisi benda
s_values = [30, 20, 15, 10, 7, 5]  # cm
labels = ['s > 2f', 's = 2f', 'f < s < 2f', 's = f', 's < f', 's << f']

print("Simulasi Lensa Cembung (f = 10 cm)")
print("-" * 50)
print(f"{'Posisi':<15} {'s (cm)':<10} {'s' (cm)':<12} {'M':<8} {'Sifat'}")
print("-" * 50)

for s, label in zip(s_values, labels):
    s_prime, M, h_prime = lens_image(f, s)
    if s_prime is None:
        print(f"{label:<15} {s:<10} {'tak hingga':<12} {'-'}")
    else:
        sifat = "Nyata, terbalik" if s_prime > 0 else "Maya, tegak"
        print(f"{label:<15} {s:<10} {s_prime:<12.1f} {M:<8.2f} {sifat}")

# Plot posisi bayangan vs posisi benda
s_plot = np.linspace(2, 50, 100)
s_prime_plot = []
for s in s_plot:
    if s != f:
        sp = 1 / (1/f - 1/s)
        s_prime_plot.append(sp)
    else:
        s_prime_plot.append(np.nan)

plt.figure(figsize=(10, 6))
plt.plot(s_plot, s_prime_plot, 'b-', linewidth=2)
plt.axhline(y=0, color='black', linewidth=0.5)
plt.axvline(x=f, color='red', linestyle='--', label=f'f = {f} cm')
plt.xlabel('Jarak benda s (cm)')
plt.ylabel('Jarak bayangan s' (cm)')
plt.title('Posisi Bayangan vs Posisi Benda untuk Lensa Cembung')
plt.grid(True, linestyle='--', alpha=0.3)
plt.legend()
plt.xlim(0, 50)
plt.ylim(-100, 100)
plt.show()
```
