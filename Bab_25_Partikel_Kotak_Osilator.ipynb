# Simulasi Fungsi Gelombang Kuantum

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import hermite, factorial

# ---------- PARTIKEL DALAM KOTAK ----------
def psi_particle_in_box(x, n, L):
    """Fungsi gelombang partikel dalam kotak 1D"""
    return np.sqrt(2/L) * np.sin(n * np.pi * x / L)

def prob_particle_in_box(x, n, L):
    """Distribusi probabilitas"""
    return psi_particle_in_box(x, n, L)**2

# ---------- OSILATOR HARMONIK KUANTUM ----------
def psi_harmonic(x, n, m, omega):
    """Fungsi gelombang osilator harmonik (skala untuk visualisasi)"""
    # Parameter disederhanakan untuk visualisasi
    H = hermite(n)
    psi = (1/np.sqrt(2**n * factorial(n) * np.sqrt(np.pi))) * np.exp(-x**2/2) * H(x)
    return psi

# Parameter untuk kotak
L = 1.0  # nm
x = np.linspace(0, L, 500)

plt.figure(figsize=(12, 8))

# Fungsi gelombang partikel dalam kotak
plt.subplot(2, 2, 1)
for n in [1, 2, 3]:
    psi = psi_particle_in_box(x, n, L)
    plt.plot(x, psi + n*0.5, label=f'n={n}')
plt.xlabel('x (nm)')
plt.ylabel('$\psi(x)$')
plt.title('Fungsi Gelombang Partikel dalam Kotak')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.3)

# Distribusi probabilitas
plt.subplot(2, 2, 2)
for n in [1, 2, 3]:
    prob = prob_particle_in_box(x, n, L)
    plt.plot(x, prob + n*0.5, label=f'n={n}')
plt.xlabel('x (nm)')
plt.ylabel('$|\psi|^2$')
plt.title('Distribusi Probabilitas')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.3)

# Osilator harmonik (visualisasi skala)
x_harm = np.linspace(-4, 4, 500)
plt.subplot(2, 2, 3)
for n in [0, 1, 2]:
    psi = psi_harmonic(x_harm, n, 1, 1)
    plt.plot(x_harm, psi + n, label=f'n={n}')
plt.xlabel('x (a.u.)')
plt.ylabel('$\psi(x)$')
plt.title('Fungsi Gelombang Osilator Harmonik')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.3)

# Energi level osilator harmonik
plt.subplot(2, 2, 4)
n_levels = np.arange(0, 6)
E_n = n_levels + 0.5
for n in n_levels:
    plt.axhline(y=E_n[n], xmin=0.1, xmax=0.9, color='blue', linewidth=1)
    plt.text(0.95, E_n[n], f'n={n}', ha='left', va='center')
plt.xlabel('x (a.u.)')
plt.ylabel('$E_n/\hbar\omega$')
plt.title('Energi Level Osilator Harmonik')
plt.xlim(0, 1)
plt.ylim(-0.5, 6)
plt.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()

print("Energi partikel dalam kotak (n=1,2,3) sebanding dengan n^2")
print("Energi osilator harmonik: E_n = (n + 1/2) hbar omega")
```
