# Simulasi Pola Interferensi dan Difraksi

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

def double_slit_intensity(theta, d, lam, N=2):
    """Intensitas celah ganda"""
    beta = np.pi * d * np.sin(theta) / lam
    I = (np.sin(N * beta / 2) / np.sin(beta / 2))**2
    I = np.where(np.abs(np.sin(beta/2)) < 1e-10, N**2, I)
    return I / N**2

def single_slit_intensity(theta, a, lam):
    """Intensitas celah tunggal"""
    beta = np.pi * a * np.sin(theta) / lam
    I = (np.sin(beta) / beta)**2
    I = np.where(np.abs(beta) < 1e-10, 1.0, I)
    return I

# Parameter
lam = 500e-9  # 500 nm
d = 0.1e-3    # 0.1 mm
a = 0.05e-3   # 0.05 mm

# Sudut
theta = np.linspace(-0.02, 0.02, 1000)

# Intensitas
I_double = double_slit_intensity(theta, d, lam)
I_single = single_slit_intensity(theta, a, lam)

# Plot
plt.figure(figsize=(12, 8))

plt.subplot(3, 1, 1)
plt.plot(theta*1000, I_double, 'b-', linewidth=1.5)
plt.xlabel('Sudut (mrad)')
plt.ylabel('Intensitas')
plt.title('Interferensi Celah Ganda')
plt.grid(True, linestyle='--', alpha=0.3)

plt.subplot(3, 1, 2)
plt.plot(theta*1000, I_single, 'r-', linewidth=1.5)
plt.xlabel('Sudut (mrad)')
plt.ylabel('Intensitas')
plt.title('Difraksi Celah Tunggal')
plt.grid(True, linestyle='--', alpha=0.3)

plt.subplot(3, 1, 3)
plt.plot(theta*1000, I_double * I_single, 'g-', linewidth=1.5)
plt.xlabel('Sudut (mrad)')
plt.ylabel('Intensitas')
plt.title('Celah Ganda + Difraksi Celah Tunggal')
plt.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()

# Tampilkan parameter
print(f"Panjang gelombang: {lam*1e9:.1f} nm")
print(f"Jarak celah: {d*1e3:.2f} mm")
print(f"Lebar celah: {a*1e3:.2f} mm")
print(f"Jarak antar terang: {lam*1e3/d:.2f} mrad")
```
