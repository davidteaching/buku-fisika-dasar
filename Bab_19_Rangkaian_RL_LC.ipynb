# Simulasi Rangkaian RL dan LC

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

# ---------- RANGKAIAN RL ----------
def rl_circuit(R, L, E, t_max, dt=0.0001):
    tau = L / R
    t = np.arange(0, t_max, dt)
    I = (E / R) * (1 - np.exp(-t / tau))
    return t, I

# ---------- RANGKAIAN LC ----------
def lc_circuit(L, C, Q0, t_max, dt=0.0001):
    omega = 1 / np.sqrt(L * C)
    t = np.arange(0, t_max, dt)
    q = Q0 * np.cos(omega * t)
    I = -Q0 * omega * np.sin(omega * t)
    # Energi
    U_E = q**2 / (2 * C)
    U_B = 0.5 * L * I**2
    U_total = U_E + U_B
    return t, q, I, U_E, U_B, U_total

# Parameter RL
R = 100  # ohm
L = 0.5  # H
E = 10   # V
t_max_rl = 0.05  # 50 ms

t_rl, I_rl = rl_circuit(R, L, E, t_max_rl)

# Parameter LC
L_lc = 0.1   # H
C_lc = 10e-6 # 10 mikroFarad
Q0 = 1e-4    # C
t_max_lc = 0.01  # 10 ms

t_lc, q_lc, I_lc, U_E, U_B, U_total = lc_circuit(L_lc, C_lc, Q0, t_max_lc)

# Plot
plt.figure(figsize=(12, 10))

# RL Charging
plt.subplot(2, 2, 1)
plt.plot(t_rl*1000, I_rl, 'b-', linewidth=2)
plt.axhline(y=E/R, color='gray', linestyle='--', label=r'$I_{\infty}=E/R$')
tau_rl = L/R
plt.axvline(x=tau_rl*1000, color='red', linestyle='--', label=r'$	au$')
plt.xlabel('Waktu (ms)')
plt.ylabel('Arus (A)')
plt.title('Pengisian Arus RL')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.3)

# LC Charge
plt.subplot(2, 2, 2)
plt.plot(t_lc*1000, q_lc*1e6, 'r-', linewidth=2)
plt.xlabel('Waktu (ms)')
plt.ylabel('Muatan ($\mu$C)')
plt.title('Osilasi Muatan LC')
plt.grid(True, linestyle='--', alpha=0.3)

# LC Current
plt.subplot(2, 2, 3)
plt.plot(t_lc*1000, I_lc*1000, 'g-', linewidth=2)
plt.xlabel('Waktu (ms)')
plt.ylabel('Arus (mA)')
plt.title('Osilasi Arus LC')
plt.grid(True, linestyle='--', alpha=0.3)

# LC Energy
plt.subplot(2, 2, 4)
plt.plot(t_lc*1000, U_E, 'r-', label='Energi listrik', linewidth=2)
plt.plot(t_lc*1000, U_B, 'b-', label='Energi magnet', linewidth=2)
plt.plot(t_lc*1000, U_total, 'k--', label='Energi total', linewidth=2)
plt.xlabel('Waktu (ms)')
plt.ylabel('Energi (J)')
plt.title('Energi dalam Rangkaian LC')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()

# Output parameter
omega_lc = 1/np.sqrt(L_lc*C_lc)
f_lc = omega_lc/(2*np.pi)
print(f"Konstanta waktu RL: tau = {L/R:.4f} s")
print(f"Frekuensi LC: f = {f_lc:.1f} Hz")
print(f"Periode LC: T = {1/f_lc:.4f} s")
```
