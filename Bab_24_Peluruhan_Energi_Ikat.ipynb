# Simulasi Peluruhan Radioaktif dan Energi Ikat

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

def decay(N0, lam, t):
    return N0 * np.exp(-lam * t)

# ---------- PELURUHAN RADIOAKTIF ----------
T_half = 5730  # tahun
lam = np.log(2) / T_half
N0 = 1000

t = np.linspace(0, 30000, 500)
N = decay(N0, lam, t)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(t, N, 'b-', linewidth=2)
plt.axhline(y=N0/2, color='gray', linestyle='--', label=f'T1/2 = {T_half} tahun')
plt.axvline(x=T_half, color='gray', linestyle='--')
plt.xlabel('Waktu (tahun)')
plt.ylabel('Jumlah inti')
plt.title('Peluruhan Radioaktif C-14')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.3)

# ---------- ENERGI IKAT PER NUKLEON ----------
A = np.array([2, 4, 12, 16, 20, 40, 56, 80, 100, 120, 140, 160, 180, 200, 220, 235])
BE_per_nucleon = np.array([1.1, 7.1, 7.7, 8.0, 8.0, 8.3, 8.8, 8.7, 8.6, 8.5, 8.4, 8.2, 8.0, 7.8, 7.6, 7.6])

plt.subplot(1, 2, 2)
plt.plot(A, BE_per_nucleon, 'ro-', linewidth=2, markersize=6)
plt.axvline(x=56, color='gray', linestyle='--', label='$A \approx 56$ (Fe)')
plt.xlabel('Nomor massa A')
plt.ylabel('Energi ikat per nukleon (MeV)')
plt.title('Kurva Energi Ikat per Nukleon')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()

# Tampilkan waktu paruh
print(f"Waktu paruh C-14: {T_half:.1f} tahun")
print(f"Konstanta peluruhan: {lam:.6f} tahun^-1")
print(f"Setelah 10.000 tahun, tersisa {N0*np.exp(-lam*10000):.1f} atom")
```
