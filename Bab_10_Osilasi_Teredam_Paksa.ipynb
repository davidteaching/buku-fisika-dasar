# Simulasi Osilasi Teredam dan Paksa

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

def damped_oscillator(m, b, k, x0, v0, t_total, dt=0.001):
    """Simulasi osilasi teredam"""
    omega0 = np.sqrt(k/m)
    beta = b/(2*m)
    omega = np.sqrt(omega0**2 - beta**2) if omega0**2 > beta**2 else 0
    
    t = np.arange(0, t_total, dt)
    x = np.zeros_like(t)
    v = np.zeros_like(t)
    x[0] = x0
    v[0] = v0
    
    for i in range(len(t)-1):
        a = -(k/m) * x[i] - (b/m) * v[i]
        v[i+1] = v[i] + a * dt
        x[i+1] = x[i] + v[i+1] * dt
    
    return t, x

def forced_oscillator(m, b, k, F0, omega_d, x0, v0, t_total, dt=0.001):
    """Simulasi osilasi paksa"""
    t = np.arange(0, t_total, dt)
    x = np.zeros_like(t)
    v = np.zeros_like(t)
    x[0] = x0
    v[0] = v0
    
    for i in range(len(t)-1):
        a = -(k/m) * x[i] - (b/m) * v[i] + (F0/m) * np.cos(omega_d * t[i])
        v[i+1] = v[i] + a * dt
        x[i+1] = x[i] + v[i+1] * dt
    
    return t, x

# Parameter
m, k = 1.0, 100.0
omega0 = np.sqrt(k/m)  # frekuensi alami = 10 rad/s

# Kasus 1: Osilasi teredam
b = 2.0  # redaman kurang kritis
t1, x1 = damped_oscillator(m, b, k, 1.0, 0.0, 10.0)

# Kasus 2: Osilasi paksa pada resonansi
F0, omega_d = 2.0, omega0
b2 = 1.0
t2, x2 = forced_oscillator(m, b2, k, F0, omega_d, 0.0, 0.0, 20.0)

# Plot
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(t1, x1, 'b-', linewidth=1.5)
plt.xlabel('Waktu (s)')
plt.ylabel('Simpangan (m)')
plt.title('Osilasi Teredam')
plt.grid(True, linestyle='--', alpha=0.6)

plt.subplot(1, 2, 2)
plt.plot(t2, x2, 'r-', linewidth=1.5)
plt.xlabel('Waktu (s)')
plt.ylabel('Simpangan (m)')
plt.title(f'Osilasi Paksa (Resonansi, $\omega_d = \omega_0 = {omega0:.1f}$ rad/s)')
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
