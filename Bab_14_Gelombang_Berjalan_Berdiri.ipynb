# Simulasi Gelombang Berjalan dan Berdiri

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

# Parameter
A = 1.0          # amplitudo
k = 2.0          # bilangan gelombang
omega = 3.0      # frekuensi sudut
x = np.linspace(0, 6, 200)

# Gelombang berjalan ke kanan
def traveling_wave(x, t):
    return A * np.sin(k * x - omega * t)

# Gelombang berdiri
def standing_wave(x, t):
    return 2 * A * np.sin(k * x) * np.cos(omega * t)

# Interferensi dua gelombang
def interference(x, t, phase_diff=0):
    y1 = A * np.sin(k * x - omega * t)
    y2 = A * np.sin(k * x - omega * t + phase_diff)
    return y1 + y2

# Buat plot untuk waktu tertentu
t = 0.5
plt.figure(figsize=(12, 8))

plt.subplot(3, 1, 1)
plt.plot(x, traveling_wave(x, t), 'b-', linewidth=2)
plt.ylim(-2.5, 2.5)
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Gelombang Berjalan (t = 0.5 s)')
plt.grid(True, linestyle='--', alpha=0.6)

plt.subplot(3, 1, 2)
plt.plot(x, standing_wave(x, t), 'r-', linewidth=2)
plt.ylim(-2.5, 2.5)
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Gelombang Berdiri (t = 0.5 s)')
plt.grid(True, linestyle='--', alpha=0.6)

plt.subplot(3, 1, 3)
plt.plot(x, interference(x, t, phase_diff=np.pi), 'g-', linewidth=2)
plt.ylim(-2.5, 2.5)
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Interferensi Destruktif (berlawanan fase)')
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
