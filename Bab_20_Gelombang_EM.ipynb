# Visualisasi Gelombang Elektromagnetik dan Spektrum

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

# ---------- GELOMBANG ELEKTROMAGNETIK ----------
def plot_em_wave():
    x = np.linspace(0, 6, 200)
    E = np.sin(2*np.pi * x)
    B = np.sin(2*np.pi * x)
    
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.plot(x, E, 'b-', linewidth=2, label=r'$ec{E}$')
    ax.plot(x, B, 'r-', linewidth=2, label=r'$ec{B}$')
    ax.set_xlabel('x (arah rambat)')
    ax.set_ylabel('Amplitudo')
    ax.set_title('Gelombang Elektromagnetik')
    ax.legend()
    ax.grid(True, linestyle='--', alpha=0.3)
    ax.axhline(y=0, color='black', linewidth=0.5)
    plt.show()

# ---------- SPEKTRUM ELEKTROMAGNETIK ----------
def plot_spectrum():
    # Frekuensi dan panjang gelombang
    names = ['Radio', 'Mikro', 'IR', 'Visible', 'UV', 'X-ray', 'Gamma']
    lambda_min = [1e-1, 1e-3, 1e-6, 4e-7, 1e-8, 1e-11, 1e-14]
    lambda_max = [1e4, 1e-1, 1e-3, 7e-7, 4e-7, 1e-8, 1e-11]
    colors = ['blue', 'cyan', 'green', 'yellow', 'orange', 'red', 'purple']
    
    plt.figure(figsize=(12, 2))
    for i, name in enumerate(names):
        plt.barh([0], width=lambda_max[i] - lambda_min[i], 
                 left=lambda_min[i], color=colors[i], label=name, height=0.6)
    
    plt.xscale('log')
    plt.xlabel('Panjang Gelombang (m)')
    plt.yticks([])
    plt.title('Spektrum Elektromagnetik')
    plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.2), ncol=7)
    plt.tight_layout()
    plt.show()

# Panggil fungsi
plot_em_wave()
plot_spectrum()

# Tampilkan konstanta
c = 3e8
mu0 = 4*np.pi*1e-7
eps0 = 1/(mu0*c**2)
print(f"Kecepatan cahaya: c = {c:.3e} m/s")
print(f"Permitivitas vakum: epsilon0 = {eps0:.3e} F/m")
print(f"Impedansi vakum: Z0 = sqrt(mu0/eps0) = {np.sqrt(mu0/eps0):.1f} ohm")
```
