# Visualisasi Kosmik: Hubble dan Komposisi Alam Semesta

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------- HUKUM HUBBLE ----------
# Data galaksi (jarak dalam Mpc, kecepatan dalam km/s)
dist = np.array([0.5, 1.0, 2.0, 3.0, 5.0, 8.0, 10.0, 15.0])
vel = np.array([35, 70, 140, 210, 350, 560, 700, 1050])  # H0 ~ 70 km/s/Mpc

# Fit linear
H0 = np.polyfit(dist, vel, 1)[0]

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(dist, vel, color='blue', s=50, label='Data galaksi')
plt.plot(dist, H0*dist, 'r-', label=f'H0 = {H0:.1f} km/s/Mpc')
plt.xlabel('Jarak (Mpc)')
plt.ylabel('Kecepatan Resesi (km/s)')
plt.title('Hukum Hubble')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.3)

# ---------- KOMPOSISI ALAM SEMESTA ----------
labels = ['Materi Biasa', 'Materi Gelap', 'Energi Gelap']
sizes = [5, 27, 68]
colors = ['blue', 'orange', 'red']
explode = (0.05, 0.05, 0.05)

plt.subplot(1, 2, 2)
plt.pie(sizes, labels=labels, colors=colors, explode=explode, 
        autopct='%1.0f%%', shadow=True, startangle=90)
plt.title('Komposisi Alam Semesta')

plt.tight_layout()
plt.show()

print(f"Konstanta Hubble H0 = {H0:.1f} km/s/Mpc")
print(f"Usia alam semesta aproksimasi: {1/(H0*3.086e19)*365*24*3600:.2e} tahun")
```
