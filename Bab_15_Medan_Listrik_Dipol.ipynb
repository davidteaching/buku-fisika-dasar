# Simulasi Medan Listrik dari Dipol

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

def electric_field(q, x, y, x0=0, y0=0):
    """Menghitung medan listrik dari muatan q di (x0,y0)"""
    k = 8.99e9
    rx = x - x0
    ry = y - y0
    r = np.sqrt(rx**2 + ry**2)
    # Hindari singularitas
    r = np.maximum(r, 0.01)
    Ex = k * q * rx / r**3
    Ey = k * q * ry / r**3
    return Ex, Ey

# Buat grid
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x, y)

# Muatan positif di (0,0)
q1 = 1e-9
Ex1, Ey1 = electric_field(q1, X, Y)

# Muatan negatif di (1,0)
q2 = -1e-9
Ex2, Ey2 = electric_field(q2, X, Y, x0=1, y0=0)

# Superposisi
Ex = Ex1 + Ex2
Ey = Ey1 + Ey2

# Plot
plt.figure(figsize=(10, 8))

# Streamplot untuk garis medan
plt.streamplot(X, Y, Ex, Ey, color='blue', linewidth=1, density=1.5)

# Muatan
plt.scatter(0, 0, s=200, color='red', label='+Q', zorder=5)
plt.scatter(1, 0, s=200, color='blue', label='-Q', zorder=5)

plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Garis-Garis Medan Listrik dari Dipol')
plt.grid(True, linestyle='--', alpha=0.3)
plt.axis('equal')
plt.legend()
plt.xlim(-2, 3)
plt.ylim(-2, 2)
plt.show()
```
