# Simulasi Aliran Fluida dalam Pipa Berubah Penampang

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

# Parameter
rho = 1000.0   # massa jenis air (kg/m^3)
g = 9.8
Q = 0.01       # debit (m^3/s)
h = 0          # ketinggian konstan

# Diameter pipa berubah sepanjang sumbu x
x = np.linspace(0, 10, 100)
D = 0.1 + 0.05 * np.sin(x * 2 * np.pi / 5)  # diameter berosilasi
A = np.pi * (D/2)**2

# Kecepatan dari kontinuitas
v = Q / A

# Tekanan dari Bernoulli (asumsi P0 = 0 di titik referensi)
P = 0.5 * rho * (v[0]**2 - v**2)  # perubahan tekanan relatif

# Plot
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8))

ax1.plot(x, D*100, 'b-', linewidth=2)
ax1.set_xlabel('Posisi (m)')
ax1.set_ylabel('Diameter (cm)')
ax1.set_title('Variasi Diameter Pipa')
ax1.grid(True, linestyle='--', alpha=0.6)

ax2.plot(x, v, 'r-', linewidth=2, label='Kecepatan')
ax2.plot(x, P/1000, 'g-', linewidth=2, label='Tekanan (kPa)')
ax2.set_xlabel('Posisi (m)')
ax2.set_ylabel('Kecepatan (m/s) / Tekanan (kPa)')
ax2.set_title('Kecepatan dan Tekanan Sepanjang Pipa')
ax2.legend()
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()

# Demonstrasi: di bagian sempit, kecepatan tinggi, tekanan rendah
idx_min = np.argmin(D)  # posisi diameter minimum
idx_max = np.argmax(D)  # posisi diameter maksimum
print(f"Diameter minimum: {D[idx_min]*100:.2f} cm, v = {v[idx_min]:.2f} m/s, P = {P[idx_min]/1000:.2f} kPa")
print(f"Diameter maksimum: {D[idx_max]*100:.2f} cm, v = {v[idx_max]:.2f} m/s, P = {P[idx_max]/1000:.2f} kPa")
```
