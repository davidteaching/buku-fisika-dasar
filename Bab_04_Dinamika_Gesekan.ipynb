# Simulasi Gerak dengan Gesekan Kinetis

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter
m = 1.0      # massa (kg)
F = 5.0      # gaya dorong (N)
mu_k = 0.2   # koefisien gesekan kinetis
g = 9.8
dt = 0.01
t_total = 10.0

# Inisialisasi
t = np.arange(0, t_total, dt)
v = np.zeros_like(t)
x = np.zeros_like(t)
v[0] = 0.0
x[0] = 0.0

# Loop simulasi
for i in range(1, len(t)):
    # Gaya gesek berlawanan arah kecepatan
    if abs(v[i-1]) > 1e-6:
        f_k = -np.sign(v[i-1]) * mu_k * m * g
    else:
        f_k = 0.0
    # Gaya total (F konstan, gesekan)
    F_net = F + f_k
    a = F_net / m
    v[i] = v[i-1] + a * dt
    x[i] = x[i-1] + v[i] * dt

# Plot
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(t, v)
plt.xlabel('Waktu (s)'); plt.ylabel('Kecepatan (m/s)')
plt.title('Kecepatan vs Waktu')
plt.grid(True)

plt.subplot(1,2,2)
plt.plot(t, x)
plt.xlabel('Waktu (s)'); plt.ylabel('Posisi (m)')
plt.title('Posisi vs Waktu')
plt.grid(True)

plt.tight_layout()
plt.show()
```
