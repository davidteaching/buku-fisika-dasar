# Simulasi Gerak Parabola

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```python
import numpy as np
import matplotlib.pyplot as plt

# Konstanta
g = 10.0      # m/s^2
v0 = 20.0     # m/s

# Fungsi untuk menghitung lintasan
def trajectory(theta_deg):
    theta = np.radians(theta_deg)
    v0x = v0 * np.cos(theta)
    v0y = v0 * np.sin(theta)
    
    # Waktu total di udara
    T = 2 * v0y / g
    t = np.linspace(0, T, 100)
    
    # Posisi
    x = v0x * t
    y = v0y * t - 0.5 * g * t**2
    return x, y

# Plot untuk beberapa sudut
angles = [30, 45, 60]
plt.figure(figsize=(10, 6))

for angle in angles:
    x, y = trajectory(angle)
    plt.plot(x, y, label=f'{angle}$^\circ$', linewidth=2)

plt.xlabel('Jarak Horizontal (m)', fontsize=12)
plt.ylabel('Ketinggian (m)', fontsize=12)
plt.title('Pengaruh Sudut Elevasi terhadap Lintasan Peluru', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.axhline(0, color='black', linewidth=0.5)
plt.show()
```
