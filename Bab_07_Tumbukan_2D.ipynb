# Simulasi Tumbukan Elastis 2D

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

def elastic_collision_2d(m1, m2, v1, v2):
    """
    Menghitung kecepatan akhir untuk tumbukan elastis 2D.
    v1, v2 adalah vektor kecepatan [vx, vy].
    """
    v1 = np.array(v1)
    v2 = np.array(v2)
    
    # Vektor normal (arah tumbukan)
    r = np.array([1, 0])  # asumsi tumbukan sepanjang sumbu x
    
    # Komponen kecepatan sepanjang garis tumbukan
    v1n = np.dot(v1, r)
    v2n = np.dot(v2, r)
    
    # Kecepatan akhir sepanjang garis tumbukan (elastis 1D)
    v1n_f = ((m1 - m2) * v1n + 2 * m2 * v2n) / (m1 + m2)
    v2n_f = ((m2 - m1) * v2n + 2 * m1 * v1n) / (m1 + m2)
    
    # Komponen tegak lurus tetap (arah tangensial)
    v1t = v1 - v1n * r
    v2t = v2 - v2n * r
    
    # Kecepatan akhir
    v1_f = v1t + v1n_f * r
    v2_f = v2t + v2n_f * r
    
    return v1_f, v2_f

# Contoh: bola 2 kg dan 4 kg
m1, m2 = 2, 4
v1 = np.array([6, 0])   # bola 1 bergerak ke kanan
v2 = np.array([0, 0])   # bola 2 diam

v1_f, v2_f = elastic_collision_2d(m1, m2, v1, v2)

print(f"Kecepatan bola 1 akhir: {v1_f} m/s")
print(f"Kecepatan bola 2 akhir: {v2_f} m/s")

# Plot vektor
plt.figure(figsize=(8, 4))
plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, 
           color='blue', width=0.02, label='Bola 1 awal')
plt.quiver(0, -0.5, v1_f[0], v1_f[1], angles='xy', scale_units='xy', scale=1, 
           color='blue', linestyle='dashed', width=0.02, label='Bola 1 akhir')
plt.quiver(0, -1.0, v2_f[0], v2_f[1], angles='xy', scale_units='xy', scale=1, 
           color='red', linestyle='dashed', width=0.02, label='Bola 2 akhir')
plt.xlim(-2, 8)
plt.ylim(-2, 2)
plt.axhline(y=0, color='black', linewidth=0.5)
plt.axvline(x=0, color='black', linewidth=0.5)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.title('Vektor Kecepatan Sebelum dan Sesudah Tumbukan Elastis')
plt.show()
```
