# Simulasi Orbit Planet dengan Metode Numerik

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

# Konstanta
G = 6.67e-11
M_sun = 2.0e30
dt = 3600 * 6  # 6 jam per langkah
t_total = 365 * 24 * 3600  # 1 tahun

# Kondisi awal Bumi (orbit mendekati lingkaran)
x0, y0 = 1.5e11, 0.0
vx0, vy0 = 0.0, 29783.0  # kecepatan orbit Bumi

# Inisialisasi
t = np.arange(0, t_total, dt)
x = np.zeros_like(t)
y = np.zeros_like(t)
vx = np.zeros_like(t)
vy = np.zeros_like(t)

x[0], y[0] = x0, y0
vx[0], vy[0] = vx0, vy0

# Simulasi (Euler-Cromer)
for i in range(len(t)-1):
    r = np.sqrt(x[i]**2 + y[i]**2)
    ax = -G * M_sun * x[i] / r**3
    ay = -G * M_sun * y[i] / r**3
    
    vx[i+1] = vx[i] + ax * dt
    vy[i+1] = vy[i] + ay * dt
    
    x[i+1] = x[i] + vx[i+1] * dt  # Euler-Cromer (gunakan kecepatan baru)
    y[i+1] = y[i] + vy[i+1] * dt

# Plot orbit
plt.figure(figsize=(8,8))
plt.plot(x, y, 'b-', linewidth=1.5, label='Orbit Bumi')
plt.plot(0, 0, 'yo', markersize=10, label='Matahari')
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Simulasi Orbit Bumi Mengelilingi Matahari')
plt.axis('equal')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()

# Periksa hukum III Kepler
T = t_total  # waktu untuk satu orbit
a = (max(x) + abs(min(x))) / 2  # sumbu semi-mayor
print(f"Periode: {T/(365*24*3600):.2f} tahun")
print(f"Sumbu semi-mayor: {a/1e11:.2f} x 10^11 m")
print(f"T^2/a^3: {T**2/a**3:.2e} s^2/m^3")
```
