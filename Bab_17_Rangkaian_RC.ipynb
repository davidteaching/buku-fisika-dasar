# Simulasi Rangkaian RC

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

# Parameter
R = 10000  # 10 kOhm
C = 100e-6  # 100 mikroFarad
E = 12.0    # Volt
tau = R * C

# Waktu
t = np.linspace(0, 5*tau, 500)

# Pengisian
V_charging = E * (1 - np.exp(-t/tau))
I_charging = (E/R) * np.exp(-t/tau)

# Pengosongan (mulai dari penuh)
V_discharging = E * np.exp(-t/tau)
I_discharging = -(E/R) * np.exp(-t/tau)

# Plot
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(12, 8))

# Tegangan pengisian
ax1.plot(t, V_charging, 'b-', linewidth=2)
ax1.axhline(y=E, color='gray', linestyle='--', alpha=0.5, label=r'$\mathcal{E}$')
ax1.axvline(x=tau, color='red', linestyle='--', alpha=0.5, label=r'$	au$')
ax1.set_xlabel('Waktu (s)')
ax1.set_ylabel('Tegangan (V)')
ax1.set_title('Pengisian Kapasitor')
ax1.legend()
ax1.grid(True, linestyle='--', alpha=0.3)

# Arus pengisian
ax2.plot(t, I_charging, 'r-', linewidth=2)
ax2.axvline(x=tau, color='red', linestyle='--', alpha=0.5)
ax2.set_xlabel('Waktu (s)')
ax2.set_ylabel('Arus (A)')
ax2.set_title('Arus Pengisian')
ax2.grid(True, linestyle='--', alpha=0.3)

# Tegangan pengosongan
ax3.plot(t, V_discharging, 'g-', linewidth=2)
ax3.axvline(x=tau, color='red', linestyle='--', alpha=0.5)
ax3.set_xlabel('Waktu (s)')
ax3.set_ylabel('Tegangan (V)')
ax3.set_title('Pengosongan Kapasitor')
ax3.grid(True, linestyle='--', alpha=0.3)

# Arus pengosongan
ax4.plot(t, I_discharging, 'm-', linewidth=2)
ax4.axvline(x=tau, color='red', linestyle='--', alpha=0.5)
ax4.set_xlabel('Waktu (s)')
ax4.set_ylabel('Arus (A)')
ax4.set_title('Arus Pengosongan')
ax4.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()

# Tampilkan nilai karakteristik
print(f"Konstanta waktu tau = RC = {tau:.3f} s")
print(f"Pada t = tau: V = {E*(1-np.exp(-1)):.2f} V (63.2%)")
print(f"Pada t = 5*tau: V = {E*(1-np.exp(-5)):.2f} V (99.3%)")
```
