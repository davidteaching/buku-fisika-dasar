# Simulasi Medan Magnet dari Kawat dan Solenoida

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```
import numpy as np
import matplotlib.pyplot as plt

def B_wire(I, x, y, x0=0, y0=0):
    """Medan magnet dari kawat lurus panjang di (x0,y0)"""
    mu0 = 4*np.pi*1e-7
    rx = x - x0
    ry = y - y0
    r = np.sqrt(rx**2 + ry**2)
    r = np.maximum(r, 0.001)
    # Arah: melingkar (untuk visualisasi)
    B = mu0 * I / (2 * np.pi * r)
    # Komponen vektor (arah melingkar)
    Bx = -B * ry / r
    By = B * rx / r
    return Bx, By

def B_solenoid(n, I, x, y, L=1.0, a=0.1):
    """Medan magnet dari solenoida ideal (aproksimasi)"""
    mu0 = 4*np.pi*1e-7
    B_inside = mu0 * n * I
    # Buat medan di dalam dan di luar
    r = np.sqrt(x**2 + y**2)
    Bx = np.where(r < a, B_inside, 0)
    By = np.zeros_like(Bx)
    return Bx, By

# Grid
x = np.linspace(-0.5, 0.5, 30)
y = np.linspace(-0.5, 0.5, 30)
X, Y = np.meshgrid(x, y)

# Kawat di (0,0) dengan arus 5 A
I1 = 5.0
Bx1, By1 = B_wire(I1, X, Y)

# Plot
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.streamplot(X, Y, Bx1, By1, color='blue', density=1.5)
plt.scatter(0, 0, s=100, color='red', label='Kawat')
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Medan Magnet dari Kawat Lurus Panjang')
plt.axis('equal')
plt.grid(True, linestyle='--', alpha=0.3)

# Solenoida
n = 500  # lilitan/m
I2 = 2.0
Bx2, By2 = B_solenoid(n, I2, X, Y)

plt.subplot(1, 2, 2)
plt.streamplot(X, Y, Bx2, By2, color='green', density=1.5)
# Gambar solenoida
plt.axhline(y=0.1, xmin=0.2, xmax=0.8, color='gray', linestyle='-', linewidth=2)
plt.axhline(y=-0.1, xmin=0.2, xmax=0.8, color='gray', linestyle='-', linewidth=2)
for x_s in np.linspace(-0.4, 0.4, 8):
    plt.plot([x_s, x_s], [-0.1, 0.1], 'gray', linewidth=1)
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Medan Magnet dalam Solenoida')
plt.axis('equal')
plt.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()

# Hitung medan di pusat solenoida
B_center = 4*np.pi*1e-7 * n * I2
print(f"Medan magnet di pusat solenoida: {B_center:.4f} T")
```
