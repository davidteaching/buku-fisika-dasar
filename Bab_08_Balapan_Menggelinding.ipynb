# Simulasi Menggelinding di Bidang Miring

Simulasi ini diambil dari buku **Catatan Kuliah Fisika Dasar** oleh David Hutasoit.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter
g = 9.8
theta = np.radians(30)  # sudut kemiringan
h = 2.0                 # ketinggian awal
L = h / np.sin(theta)   # panjang bidang miring

# Faktor k = I/(MR^2) untuk berbagai benda
bentuk = {
    'Bola Pejal': 0.4,
    'Silinder Pejal': 0.5,
    'Bola Berongga': 2/3,
    'Silinder Berongga': 1.0
}

# Hitung percepatan untuk setiap benda
waktu = {}
for nama, k in bentuk.items():
    a = g * np.sin(theta) / (1 + k)
    # Waktu tempuh: s = 0.5 a t^2 => t = sqrt(2s/a)
    t = np.sqrt(2 * L / a)
    waktu[nama] = t
    print(f"{nama}: percepatan = {a:.2f} m/s^2, waktu = {t:.2f} s")

# Plot batang perbandingan
plt.figure(figsize=(10,5))
nama_bentuk = list(waktu.keys())
waktu_tempuh = list(waktu.values())

plt.bar(nama_bentuk, waktu_tempuh, color=['blue', 'green', 'orange', 'red'])
plt.xlabel('Jenis Benda')
plt.ylabel('Waktu Tempuh (s)')
plt.title('Perbandingan Waktu Menggelinding di Bidang Miring')
plt.grid(True, axis='y', linestyle='--', alpha=0.6)

for i, v in enumerate(waktu_tempuh):
    plt.text(i, v + 0.05, f'{v:.2f}s', ha='center')

plt.show()
```
